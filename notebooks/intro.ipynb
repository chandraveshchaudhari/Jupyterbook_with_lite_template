# Welcome to the JupyterBook + JupyterLite Template

This template is designed so users can publish quickly with minimal setup.

## 1) Create Your Repository

- Click **Use this template** (or fork this repository).
- Create your own repository under your GitHub account.

## 2) Basic Project Structure

- `notebooks/`: your content pages (`.ipynb` and `.md`)
- `notebooks/_toc.yml`: page navigation and hierarchy
- `notebooks/_config.yml`: book metadata (title, author, repo)
- `.github/workflows/`: automated build and deploy to GitHub Pages
- `extensions/`: helper scripts for template behavior

## 3) Deployment Model

This template is pre-configured for GitHub Actions + GitHub Pages.
Once enabled in repository settings, every push to `main` can auto-deploy your site.
