## Template Extension Features

This template includes helper extensions that improve authoring and runtime behavior:

- **forced_jupyterlite_button.py**: Ensures stable and visible JupyterLite launch links in generated pages.
   - Run the script to create empty notebooks:
     ```bash
     python notebooks/auto_notebook_creation_using_toc.py
     ```
   - Verify notebooks in `notebooks`:
     ```bash
     ls notebooks
     ```

- **markdown_code_to_pyodide.py**: Converts specially tagged Markdown code blocks into runnable JupyterLite/Pyodide snippets.
"""python_code_block
print('Hello from JupyterLite-ready markdown!')
for i in range(3):
    print(f'Step {i}')
"""

- **auto_notebook_creation_using_toc.py**: Creates placeholder notebooks automatically from entries in `_toc.yml`.
