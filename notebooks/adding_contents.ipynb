# Add Content and Customize

This page explains how to add your own notebooks and configure project metadata.


### Editing configuration and table of content files

In **`notebooks/_config.yml`**, set:
    - `title`
    - `author`
    - `repository.url` (your GitHub username/repo)

This keeps launch links and repository buttons correct.

In **`_toc.yml`**:
   Edit `notebooks/_toc.yml` and add entries without the `.ipynb` extension.
     ```yaml
    format: jb-book
    root: intro

    chapters:
    - file: how_to_use
    title: Add Content and Customize
    - file: course_introduction
    title: Feature Showcase and Subsections
    - file: markdown_page
    title: Markdown Page Example
     ```

## Add Notebooks and content

1. Add new `.ipynb` files in the `notebooks/` folder.
2. Use descriptive file names like `chapter_name.ipynb`.
3. Keep each notebook focused on one topic for better navigation.
