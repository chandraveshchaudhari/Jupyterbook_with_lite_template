# Adding Content and Customize pages

This page explains how to add your own notebooks and configure project metadata.

### Editing Configuration and Table of Contents

Update the **`notebooks/_config.yml`** file and set:

* `title`
* `author`
* `repository.url` (your GitHub username and repository)

This ensures that repository links, launch buttons, and metadata are displayed correctly.

Update the **`notebooks/_toc.yml`** file by adding your pages **without the `.ipynb` extension**:

```yaml
format: jb-book
root: intro

chapters:
- file: adding_contents
  title: Add Content and Customize
- file: features
  title: Feature Showcase and Subsections
  sections:
      - file: section_markdown_page
        title: Section Markdown Page
```


### Add Notebooks and Content

1. Add new `.ipynb` files to the `notebooks/` folder.
2. Use clear, descriptive filenames such as `chapter_name.ipynb`.
3. Keep each notebook focused on a single topic to improve readability and navigation.
